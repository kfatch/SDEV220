```python
import multiprocessing as mp
import random
import time
from datetime import datetime as dt
import os
def process_func():
    wait_time = random.random()
    time.sleep(wait_time)
    curr_time = dt.now()
    print(curr_time)
if __name__ == "__main__":
    p1 = mp.Process(target=process_func)
    p2 = mp.Process(target=process_func)
    p3 = mp.Process(target=process_func)
    p1.start()
    p2.start()
    p3.start()
    p1.join()
    p2.join()
    p3.join()
```
