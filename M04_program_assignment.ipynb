```python
import csv
import sqlite3
import sqlalchemy
data = [
    ['title', 'author', 'year'],
    ['The Weirdstone of Brisingamen','Alan Garner','1960'],
    ['Perdido Street Station','China Miéville','2000'],
    ['Thud!','Terry Pratchett','2005'],
    ['The Spellman Files','Lisa Lutz','2007'],
    ['Small Gods','Terry Pratchett','1992']
]
file = open('books2.csv', 'w', newline='')
csvwriter = csv.writer(file)
csvwriter .writerows(data)
file.close()
```

```python
conn = sqlite3.connect('books.db')
cursor = conn.cursor()
cursor.execute('''CREATE TABLE IF NOT EXISTS books(title TEXT, author TEXT, year INTEGER)''')
cursor.executemany('''INSERT INTO books (title, author, year) VALUES (?, ?, ?)''', (data))
conn.commit()
conn.close()
```

```python
conn = sqlite3.connect('books.db')
cursor = conn.cursor()
cursor.execute('''SELECT title FROM books ORDER BY title ASC''')
title_names = cursor.fetchall()
print("Title names:", title_names)
conn.close()
```
